```python
import plotly.express as px
import pandas as pd
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
PlayerVLRRatingByDuelistDF = pd.read_csv('/content/drive/MyDrive/Player VLR Rating By Duelist/Player VLR Rating By Duelist.csv')
```

```python
fig = px.strip(PlayerVLRRatingByDuelistDF, y="VLR Rating", x="Agent", hover_name='Player',  width=1920, height=1500)
fig.update_layout(title_text='Player VLR Rating By Duelist', title_font_size= 50, title_x=0.5, title_y=0.99, plot_bgcolor = '#040A0A', paper_bgcolor = '#040A0A',
                  font=dict(color='white', size = 23),
                  xaxis=dict(ticks='outside', color='white', ticklen = 10, tickwidth = 2),
                  yaxis=dict(ticks='outside',color='white', ticklen = 10,  tickwidth = 2))

fig.update_xaxes(showgrid=False)
fig.update_yaxes(showgrid=False, range=[0,1.65])

fig.add_shape(type="line",
    x0=-1, y0=0, x1=-1, y1=2,
    line=dict(color="White",width=2)
)

fig.add_shape(type="line",
    x0=-1, y0=1, x1=7, y1=1,
    line=dict(color="White",width=2)
)

fig.update_traces(jitter=0.7, marker_size = 10)
fig.show()
```

```python
fig = px.scatter(PlayerVLRRatingByDuelistDF, y="VLR Rating", x="Agent", width=2000, height=2000)
fig.update_layout(title_text='Player VLR Rating By Duelist', title_font_size= 35, title_x=0.5, plot_bgcolor = '#040A0A', paper_bgcolor = '#040A0A',
                  font=dict(color='white', size = 23),
                  xaxis=dict(ticks='outside', color='white', ticklen = 10, tickwidth = 2),
                  yaxis=dict(ticks='outside',color='white', ticklen = 10,  tickwidth = 2))

fig.update_xaxes(showgrid=False)
fig.update_yaxes(showgrid=False, range=[0,1.8])

fig.add_shape(type="line",
    x0=-1, y0=0, x1=-1, y1=2,
    line=dict(color="White",width=2)
)

fig.add_shape(type="line",
    x0=-1, y0=1, x1=7, y1=1,
    line=dict(color="White",width=2)
)

fig.update_traces(marker_size=10, textposition = 'top center')
fig.show()
```

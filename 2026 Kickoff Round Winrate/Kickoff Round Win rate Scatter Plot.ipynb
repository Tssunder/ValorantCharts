```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
KickoffRoundWinrateDF = pd.read_csv('/content/drive/My Drive/Round Win Rate Scatter Plot/Kickoff/Round Performance Combined (All Regions).csv', skipinitialspace=True)
```

```python
plt.figure(figsize=(20, 20))
ax = sns.scatterplot(data=KickoffRoundWinrateDF,
                x="Attacking Win Percentage (%)",
                y="Defense Win Percentage (%)",
                hue_order = ["Pacific", "Americas", "EMEA", "China"],
                hue = "Region",
                edgecolor="black",
                lw=2,
                alpha=0.5
                )

# add annotations one by one with a loop
for line in range(0,KickoffRoundWinrateDF.shape[0]):
     plt.text(
          KickoffRoundWinrateDF["Attacking Win Percentage (%)"][line]+0.2,
          KickoffRoundWinrateDF["Defense Win Percentage (%)"][line],
          KickoffRoundWinrateDF["Team"][line],
          ha='left',
     )
ax.axhline(y=50, color='black', linestyle='-', linewidth=1)
ax.axvline(x=50, color='black', linestyle='-', linewidth=1)
```

```python
plt.figure(figsize=(16, 9), facecolor='#040A0A')
ax = sns.scatterplot(data=KickoffRoundWinrateDF,
                x="Attacking Win Percentage (%)",
                y="Defense Win Percentage (%)",
                edgecolor="#040A0A",
                s = 0,
                lw=0,
                alpha=1
                )

# Set axes background to #040A0A
ax.set_facecolor('#040A0A')

# Remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Make bottom and left spines white
ax.spines['bottom'].set_color('white')
ax.spines['left'].set_color('white')

# Make tick labels and axis labels white for visibility on black background
ax.tick_params(axis='x', colors='white')
ax.tick_params(axis='y', colors='white')
ax.xaxis.label.set_color('white')
ax.yaxis.label.set_color('white')
ax.legend(facecolor='#040A0A', edgecolor='white', labelcolor='white')
ax.set_title('Kickoff Round Win Rate (%)')
ax.title.set_color('white')
ax.title.set_size('20')

# Remove grid lines
ax.grid(False)

#Add Reference Lines
ax.axhline(y=50, color='white', linestyle='-', linewidth=1)
ax.axvline(x=50, color='white', linestyle='-', linewidth=1)
```

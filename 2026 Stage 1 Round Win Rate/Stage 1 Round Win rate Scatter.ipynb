```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
Stage1RoundWinrateDF = pd.read_csv('/content/drive/My Drive/Round Win Rate Scatter Plot (2026)/Stage 1/Round Performance V2 (Post Final Americas Playoffs).csv', )
```

```python
plt.figure(figsize=(20, 20))

ax = sns.scatterplot(data=Stage1RoundWinrateDF,
                x="Attacking Win Percentage (%)",
                y="Defense Win Percentage (%)",
                hue = "Region",
                edgecolor="black",
                lw=2,
                alpha=0.5
                )

# add annotations one by one with a loop
for line in range(0,Stage1RoundWinrateDF.shape[0]):
     plt.text(
          Stage1RoundWinrateDF["Attacking Win Percentage (%)"][line]+0.2,
          Stage1RoundWinrateDF["Defense Win Percentage (%)"][line],
          Stage1RoundWinrateDF["Team"][line],
          ha='left',
     )

def_average = Stage1RoundWinrateDF["Defense Win Percentage (%)"].mean()
atk_average = Stage1RoundWinrateDF["Attacking Win Percentage (%)"].mean()

ax.axhline(y=def_average, color='black', linestyle='-', linewidth=1)
ax.axvline(x=atk_average, color='black', linestyle='-', linewidth=1)

print(f'Defense Average: {def_average}')
print(f'Attack Average: {atk_average}')
```

```python
plt.figure(figsize=(16, 9), facecolor='#040A0A')
ax = sns.scatterplot(data=Stage1RoundWinrateDF,
                x="Attacking Win Percentage (%)",
                y="Defense Win Percentage (%)",
                edgecolor="#040A0A",
                s = 0,
                lw=0,
                alpha=1
                )

# Set axes background to #040A0A
ax.set_facecolor('#040A0A')

# Remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Make bottom and left spines white
ax.spines['bottom'].set_color('white')
ax.spines['left'].set_color('white')

# Make tick labels and axis labels white for visibility on black background
ax.tick_params(axis='x', colors='white')
ax.tick_params(axis='y', colors='white')
ax.xaxis.label.set_color('white')
ax.yaxis.label.set_color('white')
ax.legend(facecolor='#040A0A', edgecolor='white', labelcolor='white')
ax.set_title('Stage 1 Round Win Rate (%)')
ax.title.set_color('white')
ax.title.set_size('20')

# Remove grid lines
ax.grid(False)

#Add Reference Lines
def_average = Stage1RoundWinrateDF["Defense Win Percentage (%)"].mean()
atk_average = Stage1RoundWinrateDF["Attacking Win Percentage (%)"].mean()

ax.axhline(y=def_average, color='white', linestyle='-', linewidth=1)
ax.axvline(x=atk_average, color='white', linestyle='-', linewidth=1)

plt.xlim(37, 65)
plt.ylim(30, 65)
```
